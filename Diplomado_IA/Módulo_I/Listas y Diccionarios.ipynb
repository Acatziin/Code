# Listas

```python
# 1 - Crear Listas

nums = [1, 2, 3]  # Lista de solo números
mix = [1, "hola", True, 3.14] # Lista mixta
vacia = [] # Lista bacia
por_rango = list(range(5))      # Llena la lista con valores del 0 al 4 -> [0,1,2,3,4]
repetida = [0] * 4              # Llena la lista con el valor indicado una cantidad de veces determinada, en el ejemplo -> [0,0,0,0]
```

```python
# 2 - Indexado

a = [10, 20, 30, 40]
a[0]      # 10
a[-1]     # 40 (último)
a[1:3]    # [20,30]
a[:2]     # [10,20]
a[::2]    # [10,30]
a[::-1]   # [40,30,20,10]
```

```python
# 3 - Modificar elementos

a[1] = 99                 # [10,99,30,40]
a[1:3] = [7, 8, 9]        # [10,7,8,9,40]
a[::2] = [100, 200, 300]  # reemplaza posiciones 0,2,4. Los tamaños deben coincidir
```

```python
# 4 - Operaciones con listas

len(a) # Tamaño de la lista
10 in a # Pregunta si el elemento 10 está en la lista a. Retorna True o False
a + [5, 6] # Añade la lista [5, 6] al final de la lista a
a * 2 # Repite 2 veces la lista a -> [100, 7, 200, 9, 300, 100, 7, 200, 9, 300]
```

```python
# 5 - Agregar, insertar o eliminar

a = [1,2,3]
a.append(4)          # [1,2,3,4]
a.extend([5,6])      # [1,2,3,4,5,6]
a.insert(1, 99)      # [1,99,2,3,4,5,6]

a.pop()              # quita y regresa el último
a.pop(1)             # quita por índice
a.remove(3)          # quita primera ocurrencia del valor 3
del a[0]             # elimina por índice
a.clear()            # deja vacía
```

```python
# 6 - Búsqueda en una lista

a = ["a","b","c","b"]
a.index("b")         # Retorna el índice del elemento b (solo el primer elemento que encuentra) -> 1
a.count("b")         # Retorna la cantidad de veces que existe el elemento b -> 12
```

```python
# 7 - Ordenar y revertir

nums = [3,1,10,2]
nums.sort()                  # Ordena la lista de menor a mayor. Modifica la lista original
nums.sort(reverse=True)      # Ordena la lista de mayor a menor. Modifica la lista original

nums2 = [3,1,10,2]
ordenada = sorted(nums2)     # Ordena la lista de menor a mayor. No modifica original

nums2.reverse()              # Ordena la lista de mayor a menor. No modifica la lista original
```

```python
# 8 - Recorrer Listas

# Recorre y muestra los elementos
for x in [10, 20, 30]:
    print(x)

# Recorre y muestra los índices y elementos
for i, x in enumerate(["a","b","c"]):
    print(i, x)

# Recorre 2 listas y muestra los elementos en ambas. Deben tener el mismo tamaño
for x, y in zip([1,2,3], ["a","b","c"]):
    print(x, y)
```

```python
# 9 - Alias, copia superficial y copia profunda

a = [1,2,3]
b = a          # Alias
c = a.copy()   # Copia superficial
d = a[:]       # Copia superficial

# Realiza una copia profunda. Listas anidadas
import copy
a = [[1,2],[3,4]]
b = copy.deepcopy(a)
```

```python
# 10 - Obtener valores de la lista

a = [10, 20, 30, 40]
x, y, *resto = a     # x=10, y=20, resto=[30,40]
```

```python
# 11 - Usar las listas como una Pila (Stack)

stack = []
stack.append("A") # Añade el elemento "A" -> stack = ["A"]
stack.append("B")  # Añade el elemento "B" -> stack = ["B", "A"]
stack.pop()  # Obtiene el primer elemento "B"
```

```python
# 12 - Usar las listas como Colas (Queue)

from collections import deque

q = deque()
q.append("A") # Añade el elemento "A" -> stack = ["A"]
q.append("B") # Añade el elemento "B" -> stack = ["A", "B"]
q.popleft()  # Obtiene el elemento a la izquierda "A"
```

```python
# 13 - Listas personalizadas

cuadrados = [x*x for x in range(6)] # Crea una lista de cuadrados
pares = [x for x in range(20) if x % 2 == 0] # Crea una lista con valores pares
mayus = [s.upper() for s in ["hola", "adios"]] # Convierte las palabras en la lista a mayúsculas

mat = [[i*j for j in range(1,4)] for i in range(1,4)] # Crea una lista anidada multiplicando los índices
```

```python
# 14 - Funciones útiles

nums = [1, 5, 2]
sum(nums) # Suma los valores
min(nums) # Obtinene el mínimo
max(nums) # Obtiene el máximo

any(x > 3 for x in nums)   # True si alguno cumple
all(x > 0 for x in nums)   # True si todos cumplen
```

# Diccionarios

```python
# 1 - Crear Diccionarios

persona = {"nombre": "Ana", "edad": 20}
vacio = {}
otro = dict(nombre="Luis", edad=18)

# desde pares de elementos (tuplas)
d = dict([("a",1), ("b",2)])

# desde dos listas
claves = ["a","b"]
valores = [1,2]
d = dict(zip(claves, valores))
```

```python
# 2 - Acceder Valores

persona["nombre"]         # si no existe -> KeyError
persona.get("altura")     # None si no existe
persona.get("altura", 0)  # Si no existe da 0 como valor por defecto
```

```python
# 3 - Agregar o actualizar

persona["ciudad"] = "CDMX"     # Como la llave no existe la agrega
persona["edad"] = 21           # Como la llave existe actualiza

persona.update({"edad": 22, "tel": "555..."}) # Actualiza las llaves
```

```python
# 4 - Eliminar

del persona["tel"]
persona.pop("ciudad")          # Quita el valor con esa llave
persona.pop("noexiste", None)  # Quita el valor con esa llave y si no existe retorna None
persona.clear()
```

```python
# 5 - Recorrer Diccionarios

d = {"a":1, "b":2}

# Recorre y muestra las llaves
for k in d:
    print(k)

# Recorre y muestra los valores
for v in d.values():
    print(v)

# Recorre y muestra llaves y valores
for k, v in d.items():
    print(k, v)
```

```python
# 6 - Verificar existencia por llave

"edad" in persona
```

```python
# 7 - Listar claves y valores

list(persona.keys()) # Retorna una lista con las llaves
list(persona.values()) # Retorna una lista con los valores
list(persona.items()) # Retorna una lista con llaves y valores
```

```python
# 8 - Diccionarios anidados

alumnos = {
  "A001": {"nombre":"Ana", "calif":[10,9,8]},
  "A002": {"nombre":"Luis", "calif":[7,8,9]},
}

alumnos["A001"]["calif"].append(10)
alumnos = {
  "A001": {"nombre":"Ana", "calif":[10,9,8]},
  "A002": {"nombre":"Luis", "calif":[7,8,9]},
}

alumnos["A001"]["calif"].append(10)
```

```python
# 9 - Diccionarios personalizados

cuadrados = {x: x*x for x in range(6)} # Calcula la potencia de x
solo_pares = {k:v for k,v in cuadrados.items() if k % 2 == 0} # Llena el diccionario con solo los valores pares
```

```python
# 10 - Establecer valores por defecto

d = {}

# Sino existe, crea el calor
d.setdefault("grupo", []).append("Ana")
d.setdefault("grupo", []).append("Luis")
```

```python
# 11 - Contar Frecuencias

# Versión 1
texto = "hola hola adios"
freq = {}
for palabra in texto.split():
    freq[palabra] = freq.get(palabra, 0) + 1

# Versión 2
from collections import Counter
freq = Counter(texto.split())
```

```python
# 12 - Unir dicconarios
a = {"x": 1, "y": 2}
b = {"y": 99, "z": 3}

c = a | b      # {'x':1, 'y':99, 'z':3}
a |= b         # actualiza a con b
```

```python
# 13 - Ordenar por clave y valor
d = {"b":2, "a":3, "c":1}

por_clave = dict(sorted(d.items()))                      # por clave
por_valor = dict(sorted(d.items(), key=lambda kv: kv[1])) # por valor
```
