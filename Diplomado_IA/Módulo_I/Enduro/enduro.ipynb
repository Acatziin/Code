```python
import gymnasium as gym
import random
import ale_py # Explicitamente importar ale_py para asegurar el registro de entornos Atari

# 1. Crear el entorno
env = gym.make("ALE/Enduro-v5", render_mode="human") # Cambiado a Enduro de Atari 2600 y añadido render_mode

# 2. Reiniciar el entorno al inicio de cada episodio
observation, info = env.reset()

for step in range(500): # Aumentado el número de pasos para juegos de Atari
    # 3. Seleccionar una acción aleatoria
    action = env.action_space.sample()

    # 4. Aplicar la acción en el entorno y renderizar
    observation, reward, terminated, truncated, info = env.step(action)
    env.render() # Mostrar el entorno gráficamente

    # 5. Si el juego termina, salir
    if terminated or truncated:
        print(f"Juego terminado en el paso {step}")
        break

# 6. Cerrar el entorno
env.close()
```
