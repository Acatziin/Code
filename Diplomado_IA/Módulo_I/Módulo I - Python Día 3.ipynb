```python
# Operaciones matemáticas

print(f"Suma: {np.sum(arr1)}")
print(f"Promedio: {np.mean(arr1)}")
print(f"Desviación estándar: {np.std(arr1)}")
print(f"Máximo y mínimo: {np.max(arr1)}, {np.min(arr1)}")
print(f"Raiz cuadrada: {np.sqrt(arr1)}")
print(f"Exponencial: {np.exp(arr1)}")
print(f"Logaritmo natural: {np.log(arr1)}")
```

```python
# Indexación y Slicing

arr = np.array([[1,2,3,4], # 0
               [5,6,7,8], # 1
               [9,10,11,12]])
print(arr[0]) # Primera fila
print(arr[-1]) # Última fila
print(arr[2, 3]) # Elemento 2, 3
print(arr[:2]) # Últimas 2 filas
print(arr[:, 1:4]) # columnas de la 1 a la 3
```

```python
A = np.array([[1,2],
              [3,4]])
B = np.array([[5,6], [7,8]])

print(f"Producto Matricial (X): \n{np.dot(A,B)}")
print(f"Prodcuto Hadamard (.): \n{A*B}")
print(f"Traspuesta de A: \n{A.T}")
print(f"Determinante de A: \n{np.linalg.det(A)}")
print(f"Inversa de A: \n{np.linalg.inv(A)}")
```

```python
# Sistema de ecuaciones
coeficientes = np.array([[1,2],
              [3,4]])
constantes = np.array([5,6])

# x + 2y = 5 , 3x + 4y = 6

solucion = np.linalg.solve(coeficientes, constantes)

print(f"Solución: {solucion}")
```

```python
# Gráfico de línea

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10,100)
y = np.sin(x)

plt.figure(figsize=(10,6))
plt.plot(x,y, '-b', linewidth = 2, label = 'sin(x)')
plt.xlabel("Gráfico de la función seno", fontsize = 16)
plt.ylabel('sen(x)', fontsize = 12)
plt.grid(True, alpha = 0.3)
plt.legend()
plt.show()
```

```python
# Múltiples gráficos

plt.figure(figsize=(12,8))

# Primer subgráfico
plt.subplot(2,2,1)
plt.plot(x,np.sin(x),'r-', label= 'Sen(x)')
plt.xlabel('Sen(x)', fontname = 'Serif')
plt.title('Seno')
plt.grid(True, alpha = 0.3)
plt.legend()

# Segundo subgráfico
plt.subplot(2,2,2)
plt.plot(x,np.cos(x),'g-', label= 'Cos(x)')
plt.title('Coseno')
plt.grid(True, alpha = 0.3)
plt.legend()

# Tercer subgráfico
plt.subplot(2,2,3)
plt.plot(x,np.sin(x) * np.exp(-x/5),'b-', label= 'sin(x) * exp(-x/5)')
plt.title('Seno amortiguado')
plt.grid(True, alpha = 0.3)
plt.legend()

# Cuarto subgráfico
x2 = np.linspace(-5,5,100)
plt.subplot(2,2,4)
plt.plot(x2, x2**2,'b-', label= 'x^2')
plt.title('Parábola')
plt.grid(True, alpha = 0.3)
plt.legend()

plt.tight_layout()
plt.show()
plt.savefig('grafico.png')
```

```python
# Gráfico de dispersión

# Datos Aleatorios

np.random.seed(13)
x_scatter = np.random.randn(100)
y_scatter = 2 * x_scatter + np.random.randn(100) * 0.5

plt.figure(figsize=(10,6))
plt.scatter(x_scatter, y_scatter, alpha = 0.6, c='blue', edgecolors = 'black')
plt.title('Gráfico de dispersión')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.grid(True, alpha=0.3)

# Línea de dispersión
coef = np.polyfit(x_scatter, y_scatter,1)
poly1d_fn = np.poly1d(coef)
plt.plot(x_scatter, poly1d_fn(x_scatter), 'r-', linewidth=2, label = 'línea de tendencia')

plt.legend()
plt.show()
```

```python
# Histogramas

datos = np.random.randn(1000)

plt.figure(figsize=(10,6))
plt.hist(datos, bins= 30, alpha = 0.7, color = 'green', edgecolor = 'black')
plt.title('Histograma')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.grid(True, alpha = 0.3)

media = np.mean(datos)
plt.axvline(media, color= 'red', linestyle='--', linewidth = 2, label = f'media = {media:.2f}' )
#plt.legend()

std = np.std(datos)
plt.axvline(std, color= 'blue', linestyle='--', linewidth = 2, label = f'std = {std:.2f}' )
plt.legend()

plt.show()
```

```python
# Gráfico de barras

categorias = ["a","b","c","d"]
valores = [50,20,45,70]

plt.figure(figsize = (10,6))
bars = plt.bar(categorias, valores, color = ["red","blue","green","yellow"])

for bar in bars:
  height = bar.get_height()
  plt.text(bar.get_x() + bar.get_width()/2, height +0.5, f'{int(height)}', ha = 'center', va = 'bottom', fontsize = 10)

plt.title("Gráfico de barras")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.ylim(0,80)
plt.grid(True, alpha = 0.3, axis = 'y')
plt.legend()
plt.show()
```
