# Variables y tipos de datos

```python
# 2.1 Variables básicas
edad = 28 # Entero (int)
altura = 1.78 # Flotante
nombre = "Jorge Alberto Pacheco Senard" # Cadena (str)
es_instructor = True # booleano (bool)

print(f"Nombre: {nombre}")
print(f"Edad: {edad}")
print(f"Altura: {altura}")
print(f"Es instructor? R/ {es_instructor}")
```

```python
# edad = str(edad)
edad = "28"
edad = int(edad) # casting
```

```python
# 2.2 Tipo de datos
print(f"El tipo de dato de la variable nombre es: {type(nombre)}")
print(f"El tipo de dato de la variable edad es: {type(edad)}")
print(f"El tipo de dato de la variable altura es: {type(altura)}")
print(f"El tipo de dato de la variabe es_instructor es: {type(es_instructor)}")
```

```python
# Operaciones con números

a = 10
b = 3
c = 3.7

print(f"Suma: {a} + {b} = {a+b}")
print(f"Resta: {a} - {b} = {a-b}")
print(f"Multiplicación: {a} * {b} = {a*b}")
print(f"División: {a} / {b} = {a/b}")
print(f"División Redondeada: {a} / {b} = {round(a/b)}")
print(f"Redondeo de {c} = {round(c)}")
print(f"Truncar valor de {c} = {int(c)}")
print(f"División Entera: {a} // {b} = {a // b}")
print(f"Módulo o residuo de : {a} % {b} = {a%b}")
print(f"Potencia: {a}**{b} = {a**b}")

print(f"Resultado de: {a}+{b}*{c} = {a+b*c}")
print(f"Resultado de: ({a}+{b})*{c} = {(a+b)*c}")
```

```python
from os.path import splitext
# 2.4 Operaciones con cadenas
cadena1 = ""
cadena2 = ""

cadena1 = "Hola"
cadena2 = "Mundo"
espacio = " "
nombre = "Jorge Alberto Pacheco Senard"

print(f"Concatenación: {cadena1} + {cadena2} = {cadena1 + espacio + cadena2}")
print(f"Repetición de {cadena1} = {cadena1*3}")
print(f"Longitud de {cadena1} = {len(cadena1)}")
print(f"Longitud de {nombre} = {len(nombre)}")
print(f"Longitud de {espacio} = {len(espacio)}")
```

```python
# 2.5 Listas

numeros = [1,2,3,4,5]
frutas = ["manzana", "plátano", "naranja"]

mixta = [1,"plátano",3,"manzana"]

anidada = [mixta, numeros, frutas]

print(f"Lista de números: {numeros}")
print(f"Lista de frutas: {frutas}")
print(f"Lista mixta: {mixta}")
print(f"Lista anidada: {anidada}")
print(f"Primer elemento de la lista de números: {numeros[0]}")
print(f"Segundo elemento de la lista anidada: {anidada[1]}")
print(f"Primer elemento del segundo elemento de la lista anidada: {anidada[1][0]}")
print(f"Último numero de la lista de números: {numeros[-1]}")
print(f"Sublista de la lista de numeros: {numeros[0:4]}")
```

```python
# 2.6 Diccionarios

diccionario_estudiante = {"nombre" : "Ana",
                           "edad" : "22",
                           "carrera" : "Ingeniería",
                           "promedio" : 8.5 }

diccionario_estudiantes = {"nombres" : ["Ana", "Luis", "Julian"],
                           "edad" : [22,25,20],
                           "carrera" : "Ingeniería",
                           "promedio" : [8.5,9,9.8] }

print(f"Mostrar estudiante: {diccionario_estudiante}")
print(f"Nombre del estudiante: {diccionario_estudiante["nombre"]}")
print(f"Edad del estudiante: {diccionario_estudiante.get("edad")}")
print(f"Mostrar los estudiantes: {diccionario_estudiantes}")
print(f"Mostrar los nombres: {diccionario_estudiantes["nombres"]}")
```
