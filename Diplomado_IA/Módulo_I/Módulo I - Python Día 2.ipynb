# Clases y Programación Orientada a Objetos

```python
# 6.1 Clase Persona

class Persona:
  """Clase que representa a una persona"""

  def __init__(self) -> None:
    pass

  def __init__(self, nombre, edad):
    """Constructor de la clase persona"""
    self.nombre = nombre
    self.edad = edad

  def presentarse(self):
    """Método para presentarse"""
    return f"Hola, me llamo {self.nombre} y mi edad es de {self.edad}"

  def _aumentar_edad(self):
    """Método privado que aumenta la edad. Utilizar método cumpleanios"""
    self.edad += 1

  def cumpleanios(self):
    """Método de cumpleaños"""
    self._aumentar_edad()
    return f"¡Feliz cumpleaños a mi! Ahora tengo {self.edad} años"

# Crear instancias
persona1 = Persona("Ana", 25)
persona2 = Persona("Carlos", 30)

print(persona1.presentarse())
print(persona2.presentarse())

```

```python
print(persona1.cumpleanios())
```

```python
# 6.2 Herencia

class Estudiante(Persona):
  """Clase Estudiante que hereda de clase Persona"""

  def __init__(self, nombre = "", edad = 0, carrera = 0.0, calificaciones = []):
    super().__init__(nombre, edad)
    self.carrera = carrera
    self.calificaciones = calificaciones

  def agregar_califacion(self, calificacion):
    """Agrega una calificacion"""
    self.calificaciones.append(calificacion)

  def promedio(self):
    """Calcula el promedio"""
    if not self.calificaciones:
      return 0
    else:
      return sum(self.calificaciones)/len(self.calificaciones)

  def presentarse(self):
    return f"Hola me llamo {self.nombre}, estudio {self.carrera} y tengo {self.edad} años."

estudiante1 = Estudiante("José", 21, "Ingeniería en Computación")
estudiante1.agregar_califacion(8.5)
estudiante1.agregar_califacion(9.0)
estudiante1.agregar_califacion(7.5)

print(estudiante1.presentarse())
print(estudiante1.nombre)

print(f"Calificaciones: {estudiante1.calificaciones}")
print(f"Promedio de calificaciones: {estudiante1.promedio():.2f}")
```

```python
# 6.3 Encapsulamientoo

class CuentaBancaria:
  """Clase que representa una cuenta bancaria"""
  def __init__(self, titular, saldo_inicial):
    self.titular = titular
    self.__saldo = saldo_inicial # atributo privado

  # Setters
  def depositar(self, cantidad):
    """Depositar dinero a la cuenta si el saldo es positivo"""
    if cantidad > 0:
      self.__saldo += cantidad
      return True
    else:
      return False

  def depositar_retirar(self, cantidad):
    """Depositar dinero a la cuenta si el saldo es positivo y si es negativo retira"""
    if cantidad > 0:
      self.__saldo += cantidad
      return True
    elif cantidad < 0 and self.__saldo >= (0 - cantidad):
      self.__saldo += cantidad
      return True
    else:
      return False

  def retirar(self, cantidad):
    """Retira dinero de la cuenta"""
    if 0 < cantidad <= self.__saldo:
      self.__saldo -= cantidad
      return True
    else:
      return False

  # Getters
  def consultar_saldo(self):
    """Método para consultar saldo"""
    return self.__saldo

cuenta = CuentaBancaria("Juan", 1000)
print(f"Titular de la cuenta: {cuenta.titular}")
print(f"Saldo Inicial: {cuenta.consultar_saldo()}")
```

```python
cuenta.depositar(500)
print(f"Saldo después de depositar: {cuenta.consultar_saldo()}")
```

```python
cuenta.retirar(500)
print(f"Saldo después de retirar: {cuenta.consultar_saldo()}")
```

```python
cuenta.depositar_retirar(500)
print(f"Saldo después de depositar: {cuenta.consultar_saldo()}")
```

```python
cuenta.depositar_retirar(-500)
print(f"Saldo después de retirar: {cuenta.consultar_saldo()}")
```
