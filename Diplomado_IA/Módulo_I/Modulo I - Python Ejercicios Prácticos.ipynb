```python
# 4.1 Función que calcula el IMC

def calcular_imc(peso, altura):
  '''Calcula el IMC'''
  imc = peso / (altura**2)
  return imc

peso = float(input('Introduce tu peso en Kg: '))
altura = float(input('Introduce tu altura en metros: '))
resultado = calcular_imc(peso, altura)
print(f'Tu IMC es: {resultado:.2f}')
```

```python
# 4.2 Función que determina si es año bisiesto

def anio_bisiesto(year):
  if year % 4 == 0:
    respuesta = 'sí'
  else:
    respuesta = 'no'
  return respuesta

year = int(input('Introduce un año: '))
bisiesto = anio_bisiesto(year)
print(f'{year} {bisiesto} es un año bisiesto')
```

```python
# 4.3 Función que convierte números decimales a binarios

def dec2bin(numero):
  binario = []
  while numero > 0:
    i = int(numero % 2)
    binario.append(i)
    numero = (numero - i) / 2
  while len(binario) % 4 != 0:
    binario.append(0)

  binario.reverse()
  return binario

numero = int(input('Introduce un número a convertir: '))
resultado = dec2bin(numero)

print(f'{numero} en binario es: ', end='')
for i in range(len(resultado)):
    print(resultado[i], end = '' if (i+1)%4 else ' ')
```

```python
# 4.4 Función que calcula la media de una lista de números

def media(numeros):
  i = 0
  suma = 0

  for i in range(len(numeros)-1):
    suma += numeros[i]

  promedio = suma / len(numeros)

  return promedio

from IPython.display import clear_output

lista = []
while True:
  try:
    entrada = float(input('Introduce un número (cualquier letra o enter dos veces para salir ): '))
    lista.append(entrada)
  except ValueError:
    clear_output()
    break

resultado = media(lista)
print(f'El promedio de la lista {lista} es: {resultado:.2f}')
```

```python
# 5.1 Clase Rectángulo con métodos de área y perímetro

class Rectangulo:
  '''Clase que representa un rectángulo'''

  def __init__(self) -> None:
    pass

  def __init__(self, base, altura):
    '''Constructor del rectángulo'''
    self.base = base
    self.altura = altura

  def area(self):
    '''Método para calcular el área'''
    area = self.base * self.altura
    return f'El área del rectángulo es: {area}'

  def _aumentar_edad(self):
    '''Método privado que aumenta la edad. Utilizar método cumpleanios'''
    self.edad += 1

  def perimetro(self):
    '''Método para calcular el perímetro'''
    peri = 2 * (self.base + self.altura)
    return f'El perímetro del rectángulo es: {peri}'

# Crear instancias
rectangulo = Rectangulo(10, 25)

print(rectangulo.area())
print(rectangulo.perimetro())
```

```python
# 5.2 Método Estado de Cuenta para la clase Cuenta Bancaria. Usaré la clase mostrada por el profesor y solo haré le método: estadoCuenta

class CuentaBancaria:
  '''Clase que representa una cuenta bancaria'''

  def __init__(self, titular, saldo_inicial):
    self.titular = titular
    self.__saldo = saldo_inicial                                          # atributo privado
    self.__movimientos = []                                               # atributo privado para almacenar el historial de movimientos
    self.__movimientos.append(f'Saldo inicial: ${saldo_inicial:.2f}')     # atributo privado que registra el saldo inicial como primer movimiento

  # Setters
  def depositar_retirar(self, cantidad):
    '''Depositar dinero a la cuenta si el saldo es positivo y si es negativo retira'''
    if cantidad > 0:
      self.__saldo += cantidad
      self.__movimientos.append(f'Depósito: +${cantidad:.2f}')
      return True
    elif cantidad < 0 and self.__saldo >= abs(cantidad):
      self.__saldo += cantidad
      self.__movimientos.append(f'Retiro: -${abs(cantidad):.2f}')
      return True
    else:
      self.__movimientos.append(f'Intento fallido: {'retiro' if cantidad < 0 else 'depósito'} de ${abs(cantidad):.2f}')   # Registro de intento fallido
      return False

  # Getters
  def consultar_saldo(self):
    '''Método para consultar saldo'''
    return self.__saldo

  def estadoCuenta(self):
    '''Muestra todos los movimientos realizados y el saldo final'''
    print(f'Estado de Cuenta de {self.titular}')
    print('Historial de Movimientos:')

    for i, movimiento in enumerate(self.__movimientos, 1):
      print(f'{i}. {movimiento}')

    print(f'Saldo Final: ${self.__saldo:.2f}')

    return

# Crear una cuenta bancaria
cuenta = CuentaBancaria('Salvador Trejo', 1000.00)

# Realizar algunas operaciones
cuenta.depositar_retirar(500.00)
cuenta.depositar_retirar(-200.00)
cuenta.depositar_retirar(100.00)
cuenta.depositar_retirar(-1500.00)
cuenta.depositar_retirar(-1000.00)

# Mostrar el estado de cuenta
cuenta.estadoCuenta()
```

```python
# 5.3 Clase padre Vehículo y clase Hija Moto con método y atributos a propiados

class Vehiculo:
  '''Clase padre que representa un vehículo genérico'''

  def __init__(self, marca, modelo, anio, color, combustible='gasolina'):
    self.marca = marca
    self.modelo = modelo
    self.anio = anio
    self.color = color
    self.combustible = combustible

  # Métodos de la clase Vehículo
  def obtener_info(self):
    '''Devuelve información básica del vehículo'''
    return f'{self.marca} {self.modelo} ({self.anio}) - Color: {self.color}'

  def calcular_edad_vehiculo(anio_fabricacion):
    '''Calcula la edad del vehículo basado en el año actual'''
    from datetime import datetime
    anio_actual = datetime.now().year
    return anio_actual - anio_fabricacion

class Moto(Vehiculo):
  '''Clase hija que representa una motocicleta que hereda de Vehiculo'''

  def __init__(self, marca, modelo, anio, color, cilindrada, tipo='deportiva', combustible='gasolina', tiene_casco_integrado=False):
    super().__init__(marca, modelo, anio, color, combustible)
    self.cilindrada = cilindrada
    self.tipo = tipo
    self.tiene_casco_integrado = tiene_casco_integrado

  # Métodos específicos de Moto
  def obtener_info(self):
    '''Devuelve información detallada de la moto'''
    info_base = super().obtener_info()
    if self.tiene_casco_integrado:
      casco_info = 'Con casco incluido'
    else:
      casco_info = 'Sin casco incluido'
      return (f'{info_base} | Tipo: {self.tipo} | Cilindrada: {self.cilindrada}cc | {casco_info}')

  def __str__(self):
    '''Representación en string de la moto'''
    return self.obtener_info()

auto = Vehiculo('Toyota', 'Corolla', 2020, 'Rojo')
print(auto.obtener_info())
print(f'Edad del auto: {Vehiculo.calcular_edad_vehiculo(auto.anio)} años')

print('\nCreando una motocicleta: ')
mi_moto = Moto(marca='Honda', modelo='CBR350', anio=2022, color='Azul', cilindrada=350, tipo='Cafe Racer', tiene_casco_integrado=True)
print(mi_moto.obtener_info())
print(f'Edad de la moto: {Moto.calcular_edad_vehiculo(mi_moto.anio)} años')

print('\nCreando otra motocicleta: ')
moto_custom = Moto(marca='Harley-Davidson', modelo='Softail', anio=2018, color='Negro', cilindrada=1800, tipo='Custom', tiene_casco_integrado=False)
print(moto_custom.obtener_info())
print(f'Edad de la moto: {Moto.calcular_edad_vehiculo(moto_custom.anio)} años')
```

```python
# 5.4 Clase Libro con atributos y métodos apropiados
import builtins # En una celda de arriba utilicé sum como variable y no puede limpiarla, por lo que creaba error con la función sum

class Libro:
  '''Clase que representa un libro'''
  def __init__(self, titulo, autor, isbn, paginas, genero='Sin género especificado', editorial='Desconocida'):
    self.titulo = titulo
    self.autor = autor
    self.isbn = isbn
    self.paginas = paginas
    self.genero = genero
    self.editorial = editorial
    self.calificaciones = []

  def calificar(self, puntuacion, comentario=''):
    '''Califica el libro con una puntuación y comentario'''
    if 1 <= puntuacion <= 5:
      calificacion = {'puntuacion': puntuacion, 'comentario': comentario, 'fecha': self._obtener_fecha_actual()}
      self.calificaciones.append(calificacion)
      print(f'Calificación registrada para {self.titulo}: {puntuacion}/5')
      return True
    else:
      print('La puntuación debe estar entre 1 y 5')
      return False

  def obtener_promedio_calificaciones(self):
    '''Calcula el promedio de calificaciones del libro'''
    if not self.calificaciones:
      return 0.0
    promedio = builtins.sum(cal['puntuacion'] for cal in self.calificaciones) / len(self.calificaciones)
    return round(promedio, 2)

  def info_detallada(self):
    '''Muestra información detallada del libro'''
    promedio = self.obtener_promedio_calificaciones()
    info = f'''\nINFORMACIÓN DEL LIBRO
    Título: {self.titulo}
    Autor: {self.autor}
    ISBN: {self.isbn}
    Páginas: {self.paginas}
    Género: {self.genero}
    Editorial: {self.editorial}
    Calificaciones: {promedio}/5*({len(self.calificaciones)} calificaciones)'''

    if self.calificaciones:
      info += '\nÚltimas calificaciones:'
      for i, cal in enumerate(self.calificaciones[-3:], 1):  # Mostrar las últimas 3
        fecha = cal['fecha']
        comentario = f' - {cal['comentario']}' if cal['comentario'] else ''
        info += f'\n        {i}. {cal['puntuacion']}/5 ({fecha}){comentario}'

      return info

  def resumen_ejecutivo(self):
    '''Muestra un resumen ejecutivo del libro'''
    return f'{self.titulo} de {self.autor} - {self.paginas} páginas - {self.genero}'

  def _obtener_fecha_actual(self):
    '''Método privado para obtener la fecha actual'''
    from datetime import datetime
    return datetime.now().strftime('%d/%m/%Y')

  def __str__(self):
    '''Representación en string del libro'''
    return f'Libro: {self.titulo} | Autor: {self.autor} | ISBN: {self.isbn} | Páginas: {self.paginas}'

# Crear algunos libros
libro1 = Libro(titulo='Cien años de soledad', autor='Gabriel García Márquez', isbn='978-0307474728', paginas=417, genero='Realismo mágico', editorial='Sudamericana')
libro2 = Libro(titulo='1984', autor='George Orwell', isbn='978-0451524935', paginas=328, genero='Ciencia ficción distópica', editorial='Secker & Warburg')
libro3 = Libro(titulo='El principito', autor='Antoine de Saint-Exupéry', isbn='978-0156012195', paginas=96, genero='Literatura infantil filosófica', editorial='Reynal & Hitchcock')

print('\n1. Libros disponibles:')
print(libro1)
print(libro2)
print(libro3)

print('\n2. Calificaciones:')
libro1.calificar(5, 'Una obra maestra de la literatura latinoamericana')
libro1.calificar(3.5, 'Muy bueno, pero algo confuso con los nombres')
libro1.calificar(3, 'Interesante, pero no es mi estilo')

libro2.calificar(5, 'Increíblemente visionario')
libro2.calificar(4.8, 'Más relevante que nunca')

libro3.calificar(5, 'Es una obra maravillosa, creía que era una novela infantil pero es para todos los públicos')
libro3.calificar(4.5, 'Imprescindible')

print('\n3. Promedio de calificaciones:')
print(f'Promedio de {libro1.titulo} : {libro1.obtener_promedio_calificaciones()}/5')
print(f'Promedio de {libro2.titulo} : {libro2.obtener_promedio_calificaciones()}/5')
print(f'Promedio de {libro3.titulo} : {libro3.obtener_promedio_calificaciones()}/5')

print('\n4. Resumen Ejecutivo:')
print(libro1.resumen_ejecutivo())
print(libro2.resumen_ejecutivo())

print('\n5. Inforamción Detallada:')
print(libro1.info_detallada())
print(libro3.info_detallada())
```

```python
# 6.3 Resuelve el sistema de ecuaciones lineales:
#     3𝑥 +5𝑦 + 2𝑧  = 15
#     5𝑥 + 𝑦 + 𝑧 = 8
#     𝑥 – 𝑦 + 5𝑧 = 4

coeficientes = np.array([[3,5,2],
                         [5,1,1],
                         [1,-1,5]])

constantes = np.array([15,8,4])

solucion = np.linalg.solve(coeficientes, constantes)

print(f"Solución: \nx = {solucion[0]:.0f}\ny = {solucion[1]:.0f}\nz = {solucion[2]:.0f}")
```

```python
# 7.1 Gráfico de la función coseno

x = np.linspace(0,10,100)
y = np.cos(x)

plt.figure(figsize=(10,6))
plt.plot(x,y, 'maroon', linewidth = 2, label = 'f(x) = cos x')
plt.title('Gráfico de la función coseno')
plt.xlabel('x', fontsize = 12)
plt.ylabel('f(x)', fontsize = 12)
plt.grid(True, alpha = 0.3)
plt.legend()
plt.show()
```

```python
# 7.2 Histograma de datos aleatorios normalmente distribuidos

datos = np.random.randn(1000)

plt.figure(figsize=(10,6))
plt.hist(datos, bins= 30, alpha = 0.7, color = 'maroon', edgecolor = 'black')
plt.title('Histograma')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.grid(True, alpha = 0.3)
plt.legend()

plt.show()
```

```python
# 7.3 Gráfico de dispersión con línea de tendencia

np.random.seed(33)
peso = np.sort(np.random.uniform(50,120,100))
altura = np.sort(np.random.uniform(1.45, 2.14, 100))

plt.figure(figsize=(10,6))
plt.scatter(peso, altura, alpha = 0.6, c='green', edgecolors = 'black')
plt.title('Gráfico de dispersión')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.grid(True, alpha=0.3)

# Línea de dispersión
coef = np.polyfit(peso, altura, 1)
poly1d_fn = np.poly1d(coef)
plt.plot(peso, poly1d_fn(peso), 'maroon', linewidth=2, label = 'línea de tendencia')

plt.legend()
plt.show()
```
