# TAREA: Preprocesamiento de texto en español

```python
# 1. Imports necesarios
import nltk
import re
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
```

```python
# 2. Descarga de recursos
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# 3. Función auxiliar para emojis
# Puedes dejar esta parte dada para que la tarea se enfoque
# en el pipeline principal y no en construir todo desde cero.
def reemplazar_emojis(texto):
    mapa_emojis = {
        "😡": " emoji_enojo ",
        "😠": " emoji_molestia ",
        "😊": " emoji_felicidad ",
        "😀": " emoji_alegria ",
        "😍": " emoji_amor ",
        "😢": " emoji_tristeza ",
        "😭": " emoji_llanto ",
        "😱": " emoji_sorpresa ",
        "🤢": " emoji_asco ",
        "🍕": " emoji_pizza ",
        "🐶": " emoji_perro "
    }

    for emoji, etiqueta in mapa_emojis.items():
        texto = texto.replace(emoji, etiqueta)

    return texto
```

```python
# 4. Textos de prueba
resena_1 = "La pizza estaba <b>INCREÍBLE</b>. 🍕 Llegó en 20 minutos. http://pizza.com"
resena_2 = "¿Alguien sabe si abren el domingo? Quiero llevar a mis 2 perros. 🐶"
resena_3 = "No me gustó el servicio, pero la decoración no estaba mal."
```

```python
# 5. Función a completar por el alumno
def limpiar_resena(texto):
    """
    Implementa una función de preprocesamiento en español.

    Debe incluir al menos:
    - quitar HTML
    - quitar URLs
    - pasar a minúsculas
    - reemplazar emojis por etiquetas
    - quitar símbolos extraños
    - quitar puntuación
    - decidir qué hacer con números
    - quitar espacios extra
    - tokenizar
    - quitar stopwords, conservando negaciones
    - aplicar stemming
    """

    # 1. Quitar HTML
    texto_sin_html = re.sub(r'<.*?>', ' ', texto)

    # 2. Quitar URLs
    texto_sin_urls = re.sub(r'http\S+|www\S+', ' ', texto_sin_html)

    # 3. Pasar a minúsculas
    texto_minusculas = texto_sin_urls.lower()

    # 4. Reemplazar emojis
    texto_sin_emojis = reemplazar_emojis(texto_minusculas)

    # 5. Quitar símbolos raros
    texto_limpio_parcial = re.sub(r'[^\w\sáéíóúüñ¡!¿?.,]', ' ', texto_sin_emojis)

    # 6. Quitar puntuación
    texto_sin_puntuacion = texto_limpio_parcial.translate(str.maketrans('', '', string.punctuation + '¡¿'))

    # 7. Quitar números
    texto_sin_numeros = re.sub(r'\d+', ' ', texto_sin_puntuacion)

    # 8. Quitar espacios extra
    texto_normalizado = re.sub(r'\s+', ' ', texto_sin_numeros).strip()

    # 9. Tokenizar
    tokens = word_tokenize(texto_normalizado, language='spanish')

    # 10. Quitar stopwords conservando negaciones
    stop_words_es = set(stopwords.words('spanish'))
    stop_words_es = stop_words_es - {'no', 'ni', 'nunca'}    #conservar negaciones

    tokens_sin_stopwords_es = [palabra for palabra in tokens if palabra not in stop_words_es]

    # 11. Aplicar stemming
    stemmer = SnowballStemmer("spanish")

    tokens_stem = [stemmer.stem(palbra) for palbra in tokens_sin_stopwords_es]

    return tokens, tokens_sin_stopwords_es, tokens_stem
```

```python
# 6. Pruebas
# Descomenta estas líneas cuando termines tu función:
print('Reseña 1:')
print('Tokens:', limpiar_resena(resena_1)[0])
print('Tokens sin StopWords:', limpiar_resena(resena_1)[1])
print('Tokens Stemmizados:', limpiar_resena(resena_1)[2])

print('\nReseña 2:')
print('Tokens:', limpiar_resena(resena_2)[0])
print('Tokens sin StopWords:', limpiar_resena(resena_2)[1])
print('Tokens Stemmizados:', limpiar_resena(resena_2)[2])

print('\nReseña 3:')
print('Tokens:', limpiar_resena(resena_3)[0])
print('Tokens sin StopWords:', limpiar_resena(resena_3)[1])
print('Tokens Stemmizados:', limpiar_resena(resena_3)[2])
```
