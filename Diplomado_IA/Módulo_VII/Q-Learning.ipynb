```python
import gymnasium as gym
import numpy as np
from random import randint
```

```python
env= gym.make("MountainCar-v0")
```

[ambiente](https://gymnasium.farama.org/environments/classic_control/mountain_car/)

```python
def discretizar(valor):
    aux = ((valor - env.observation_space.low)/(env.observation_space.high-env.observation_space.low))*20
    return tuple(aux.astype(np.int32))
```

```python
q_table = np.random.uniform(low=-1, high=1, size=[20,20,3])
```

```python
#print(discretizar(env.reset()[0]))
#env.close()
```

```python
alfa = 0.1 
gamma = 0.95
epsilon = 2
episodios = 5000
#acReward=0
lista_recompenzas = []
for episodio in range(episodios):
    estado=discretizar(env.reset()[0])
    final= False
    f1 = False
    f2 = False
    recompensa_total = 0
    while not final:
        if randint(0,10)>epsilon:
            accion = np.argmax(q_table[estado])
        else:
            accion =randint(0,2)
            
        nuevo_estado,recompensa,f1,f2,info = env.step(accion)
        q_table[estado][accion] = q_table[estado][accion] + alfa*(recompensa+gamma*np.max(q_table[discretizar(nuevo_estado)])-q_table[estado][accion])
        estado= discretizar(nuevo_estado)
        recompensa_total+=recompensa
        #acReward+=recompensa
        final= f1 or f2
    lista_recompenzas.append(recompensa_total)
    if (episodio+1)%100==0:
        print("Episodio: "+str(episodio)+"| Recompensa promedio: "+str(np.mean(lista_recompenzas)))
        #print("Episodio: "+str(episodio)+"| Recompensa promedio: "+str(acReward/100))
        #acReward=0
env.close()
```

```python
import gymnasium as gym
import numpy as np
from random import randint
```

```python
env= gym.make("Pendulum-v1")
```

```python
def discretizar(valor):
    aux = ((valor - env.observation_space.low)/(env.observation_space.high-env.observation_space.low))*19
    return tuple(aux.astype(np.int32))
```

```python
#print(discretizar(env.reset()[0]))
#env.close()
```

```python
def continous(indice):
  if indice == 0:
    return np.array([-1.99], dtype=np.float32)
  elif indice == 1:
    return np.array([-1.0], dtype=np.float32)
  elif indice == 2:
    return np.array([0.0], dtype=np.float32)
  elif indice == 3:
    return np.array([1.0], dtype=np.float32)
  else:
    return np.array([1.99], dtype=np.float32)
```

```python
q_table = np.random.uniform(low=-1, high=1, size=[20,20,20,5])
```

```python
alfa = 0.1
gamma = 0.95
episodios = 5000
epsilon=2
lista_recompenzas = []
for episodio in range(episodios):
    estado=discretizar(env.reset()[0])
    final= False
    f1 = False
    f2 = False
    recompensa_total = 0
    while not final:
        if randint(0,10)>epsilon:
            accion = np.argmax(q_table[estado])
        else:
            accion = randint(0,4)

        nuevo_estado,recompensa,f1,f2,info = env.step(continous(accion))
        q_table[estado][accion] = q_table[estado][accion] + alfa*(recompensa+gamma*np.max(q_table[discretizar(nuevo_estado)])-q_table[estado][accion])
        estado= discretizar(nuevo_estado)
        recompensa_total+=recompensa
        final= f1 or f2
    lista_recompenzas.append(recompensa_total)
    if (episodio+1)%100==0:
        print("Episodio: "+str(episodio)+"| Recompensa promedio: "+str(np.mean(lista_recompenzas)))

env.close()
```

```python
env= gym.make("Pendulum-v1",render_mode="human")
```

```python
estado=discretizar(env.reset()[0])
final= False
f1 = False
f2 = False
recompensa_total = 0
while not final:
    
    accion = np.argmax(q_table[estado])
      
    nuevo_estado,recompensa,f1,f2,info = env.step(continous(accion))
    estado= discretizar(nuevo_estado)
  
    final= f1 or f2
env.close()
```
