```python
# ============================================================
# PREGUNTA 1
# ============================================================
#
# Explica con tus propias palabras:
#
# ¿Qué diferencia existe entre una red neuronal tradicional
# feedforward y una RNN?
#
# Menciona:
# - memoria
# - dependencia temporal
# - procesamiento secuencial
#
# RESPUESTA:
#
```

La principal diferencia entre una red neuronal tradicional y una Red Neuronal Recurrente está en cómo gestionan el flujo de información y el contexto temporal de los datos.

Memoria: Mientras que las redes feedforward sufren de "amnesia severa" porque la información fluye en una sola dirección sin guardar rastro del pasado, las RNN introducen un bucle interno o conexión retroalimentada que permite que la información persista. Este mecanismo genera un estado oculto (ht​), que funciona como una "memoria matemática" o resumen comprimido de lo que la red ha visto en pasos anteriores.

Dependencia temporal: En una red tradicional, el resultado de una entrada no afecta a la siguiente (Suposición de Markov). En una RNN, existe una dependencia temporal estricta: el output en el tiempo t depende tanto de la entrada actual (xt​) como del estado pasado (ht−1​). Esto permite que la red conecte contextos distantes, algo vital para entender tareas como el lenguaje natural, donde el significado de una palabra depende de las que aparecieron antes.

Procesamiento secuencial: Las redes tradicionales procesan entradas de tamaño fijo de forma independiente, asumiendo que no hay relación entre un ejemplo y el siguiente. En cambio, las RNN están diseñadas específicamente para el procesamiento secuencial; operan iterando a través de los elementos de una secuencia, alineando las posiciones de los datos con pasos de tiempo computacionales.

En resumen, mientras la red tradicional es un lector que olvida cada página al terminarla, la RNN es un sistema dinámico que mantiene un historial continuo para interpretar la secuencia completa.

```python
# ============================================================
# PREGUNTA 3
# ============================================================
#
# ¿Por qué una RNN simple puede olvidar información lejana?
#
# Explica:
# - gradiente desvaneciente
# - multiplicaciones sucesivas
# - saturación de tanh o sigmoid
#
# RESPUESTA:
#
```

Las RNN simples (como la de Elman) olvidan información lejana principalmente debido a las multiplicaciones sucesivas de sus pesos durante el proceso de Backpropagation Through Time. Al reutilizar la misma matriz de pesos en cada paso, el error debe viajar hacia atrás multiplicándose repetidamente por sí mismo; si estos valores son pequeños, la señal se debilita de forma exponencial hasta desaparecer. Este fenómeno, conocido como gradiente desvaneciente, crea un "abismo del tiempo" que impide que la información del inicio de una secuencia larga influya en el aprendizaje actual, provocando una amnesia estructural en el modelo.
A esto contribuye la saturación de tanh o sigmoid, funciones utilizadas para comprimir la información en la red. Cuando estas funciones operan en sus valores máximos o mínimos, entran en zonas de saturación donde sus curvas son casi planas y sus derivadas son prácticamente cero. Al multiplicar estas derivadas tan pequeñas a lo largo de muchos pasos de tiempo, la señal del error se "aplasta" instantáneamente, lo que hace que la red solo sea capaz de conectar el pasado inmediato con el presente y falle al intentar retener contextos profundos.

Las redes LSTM se diseñaron específicamente para resolver el problema del gradiente desvaneciente, y la clave de la solución reside en las compuertas "gates", que funcionan como reguladores de flujo para proteger la memoria. La LSTM introduce un Cell State o "cinta transportadora" donde la información viaja preservando su gradiente casi intacto, gracias a que el estado se actualiza mediante sumas aditivas en lugar de multiplicaciones repetitivas. Las compuertas usan funciones sigmoides para tomar decisiones quirúrgicas. Así, estas "válvulas" matemáticas permiten que la red elija deliberadamente qué olvidar y qué recordar, logrando una memoria selectiva y de largo plazo.

Mientras que ReLU ayuda a evitar el gradiente desvaneciente en redes feed-forward, en las recurrentes como la LSTM, la tanh y la sigmoid son las que garantizan que el flujo de memoria sea controlable y no se desborde, por eso normalmente ReLu no se usa como activación principal, y por un par de razones críticas de diseño:

1. Naturaleza de las gates: Las compuertas (Forget, Input y Output) necesitan funcionar como "válvulas" precisas que dejen pasar un porcentaje de información (de 0% a 100%). Para esto, la función sigmoid es indispensable, ya que entrega valores estrictamente entre 0 y 1. ReLU, al no tener un límite superior, no puede representar una proporción o probabilidad, lo que rompería el mecanismo de control de flujo de la celda.

2. Estabilidad del Cell State: La LSTM genera nuevos "candidatos de conocimiento" (C~t​) que se añaden a la cinta transportadora mediante sumas aditivas. Se utiliza tanh como activación para estos candidatos porque actúa como un "regulador de presión" que acota los valores entre -1 y 1. Si usáramos ReLU (que es ilimitada hacia el infinito), los valores en el Cell State podrían crecer desproporcionadamente paso tras paso debido a las sumas constantes.

La diferencia fundamental reside en la simplificación estructural: mientras que la LSTM utiliza tres compuertas (Forget, Input y Output) y mantiene dos estados separados (celda y oculto), la GRU reduce el diseño a solo dos compuertas (Update y Reset) y fusiona todo en un único canal unificado. En la GRU, la Update Gate realiza una función doble al decidir simultáneamente qué proporción de la memoria antigua se debe descartar y qué información nueva se debe integrar, optimizando el flujo de datos.

La GRU tiene menos parámetros, lo que se traduce directamente en un entrenamiento más veloz y un menor consumo de memoria GPU. El problema del cuello de botella ocurre porque el Encoder está obligado a comprimir toda la riqueza semántica de una oración (sin importar si tiene 5 o 500 palabras) en un único Vector de Contexto (C) de tamaño fijo. Matemáticamente, es imposible resumir secuencias masivas en un solo número o vector sin perder detalles finos.
Esta limitación provoca lo que llamamos amnesia neuronal, donde el Encoder se sobrecarga al intentar entender el significado y comprimirlo a la fuerza al mismo tiempo. Como el proceso es puramente secuencial, para cuando la red llega al final de una frase larga, la información del inicio ya se ha desvanecido o ha sido sobrescrita, resultando en una caída drmática del rendimiento en secuencias de más de 20 palabras. En resumen, cualquier pérdida de información en este "tubo matemático" es permanente y el Decoder solo puede "adivinar".


```python
# ============================================================
# PREGUNTA 10
# ============================================================
#
# En atención:
#
# ¿Qué representa cada uno?
#
# a) Query
# b) Key
# c) Value
#
# RESPUESTA:
#
```

Representan un sistema de búsqueda (Soft Lookup) para que el modelo no tenga que comprimir todo el pasado en un solo vector.
Según la analogía del buscador:
- Query (Q) – ¿Qué busco?
Representa la consulta o necesidad actual del Decoder en un paso de tiempo específico.
- Key (K) – ¿Qué tienes?
Son las etiquetas descriptivas o índices de cada palabra procesada por el Encoder. Funcionan como metadatos que sirven para calcular la relevancia matemática.
- Value (V) – El Contenido.
Representa la información semántica pura o el "tesoro" de conocimiento que reside en cada estado del Encoder. Una vez que se decide qué palabras son importantes (haciendo match entre Q y K), el sistema extrae el contenido de los Values para construir el vector de contexto dinámico.
