## Instalar transformers torch openai

```python
# Simular como un modelo de lenguaje predice la siguiente palabra

contexto = "La inteligencia artificial"

posibles_palabras = {
    "cambiará" : 0.45,
    "mejorará" : 0.25,
    "transformará": 0.20,
    "desaparecerá:": 0.10
}


print(f"Contexto: {contexto}")
```

```python
print("\n Probabilidades de la siguiente palabra")

for palabra, probabilidad in posibles_palabras.items():
  print(f"{palabra}: {probabilidad}")
```

```python
siguiente_palabra = max(posibles_palabras, key=posibles_palabras.get)

print(f"Palabra seleccionada: {siguiente_palabra}")
```

```python
print(f"Texto generado: {contexto} {siguiente_palabra}")
```

Predicción con temperatura

```python
import random

posibles_palabras = {
    "cambiará" : 0.45,
    "mejorará" : 0.25,
    "transformará": 0.20,
    "desaparecerá:": 0.10
}

palabras = list(posibles_palabras.keys())
probabilidades = list(posibles_palabras.values())
```

```python
for i in range(15):
  seleccion = random.choices(palabras, probabilidades, k = 1)
  print(contexto + " " + seleccion[0])
```

```python
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

texto = "La inteligencia artificial cambiará al mundo."
```

```python
texto_reconstruido = tokenizer.decode(ids)

print(f"Texto Original: {texto}")
print(f"Texto Reconstruído: {texto_reconstruido}")
```

Observar como GPT2 calcula los tokens más probables después de un texto

```python
import torch
from transformers import GPT2Tokenizer, GPT2LMHeadModel

tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')
```

```python
texto = "The future of artificial intelligence is"

inputs = tokenizer(texto, return_tensors = "pt")

with torch.no_grad():
  outputs = model(**inputs)

logits = outputs.logits

ultimo_token_logits = logits[0, -1,:]

top_k = 100

valores, indices = torch.topk(ultimo_token_logits, k = top_k)
```

```python
print(f"Texto: {texto}")

print("Tokens más probables:")

for valor, indice in zip(valores, indices):
  token = tokenizer.decode([indice])
  probabilidad = torch.softmax(ultimo_token_logits, dim = 0)[indice].item()
  print(f"{token} -> {probabilidad:.4f}")
```

# GPT1 vs GT2

```python
from transformers import pipeline

prompt = "Machine learning allows computers to"
```

```python
modelos = {
    "GPT-1": "openai-gpt",
    "GPT-2": "gpt2"
}

for nombre, modelo in modelos.items():
  print(f"\nModelo: {nombre}")

  generador = pipeline("text-generation", model = modelo)

  resultado = generador(
      prompt,
      max_length = 60,
      do_sample = True,
      top_k = 50,
      top_p = 0.95,
      temperature = 0.7,
      num_return_sequences = 1
  )

  print(resultado[0]["generated_text"])
```
